**Formateo de archivos JSON (Código 1):**

    *   Ejecuta el Código 1 (`formatear_json_en_ruta`) para formatear los archivos JSON copiados en el paso anterior. Esto aplicará un formato legible a los archivos JSON, lo que facilitará su posterior manipulación y modificación.

```python
import json
import os

def formatear_json_en_ruta(ruta_carpeta, archivos_a_formatear):
    """
    Formatea archivos JSON específicos en una carpeta.

    Args:
        ruta_carpeta (str): La ruta a la carpeta que contiene los archivos JSON.
        archivos_a_formatear (list): Una lista de nombres de archivos (con extensión .json)
                                     que se van a formatear. Si está vacía, se formatearán
                                     todos los archivos .json en la carpeta.
    """

    try:
        # Verifica si la ruta existe
        if not os.path.exists(ruta_carpeta):
            raise FileNotFoundError(f"La ruta '{ruta_carpeta}' no existe.")

        # Itera sobre todos los archivos en la carpeta
        for nombre_archivo in os.listdir(ruta_carpeta):
            if nombre_archivo.endswith('.json'):
                # Verifica si el archivo está en la lista de archivos a formatear
                if archivos_a_formatear and nombre_archivo not in archivos_a_formatear:
                    continue  # Saltar este archivo si no está en la lista

                ruta_archivo = os.path.join(ruta_carpeta, nombre_archivo)
                print(f'Procesando archivo: {ruta_archivo}')

                try:
                    # Abre el archivo JSON en modo lectura
                    with open(ruta_archivo, 'r', encoding='utf-8') as f:
                        datos_json = json.load(f)

                    # Guarda los cambios en el archivo JSON con formato
                    with open(ruta_archivo, 'w', encoding='utf-8') as f:
                        json.dump(datos_json, f, indent=4, ensure_ascii=False)  # indent=4 para formato, ensure_ascii=False para caracteres especiales

                except json.JSONDecodeError:
                    print(f"Error: El archivo '{nombre_archivo}' no es un archivo JSON válido.")
                except Exception as e:
                    print(f"Error al procesar '{nombre_archivo}': {e}")

    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(f"Error general: {e}")


# Ejemplo de uso:
ruta_carpeta = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS BASE'

# Lista de archivos específicos a formatear (opcional). Si no se proporciona, se formatean todos los .json
archivos_a_formatear = []  # Ejemplo: ['archivo1.json', 'archivo2.json']

formatear_json_en_ruta(ruta_carpeta, archivos_a_formatear)
```

**Copia de archivos a carpetas (Código 2):**

    *   Ejecuta el Código 2 (`copiar_archivos_a_carpetas`) para copiar los archivos JSON desde la carpeta base a las carpetas individuales en la carpeta destino. Esto te permitirá trabajar con copias de los archivos originales y preservar los originales sin modificaciones.


```python
import os
import shutil

def copiar_archivos_a_carpetas(ruta_base, ruta_destino):
    # ... (documentación de la función)

    try:
        # Verifica si las rutas existen
        if not os.path.exists(ruta_base):
            raise FileNotFoundError(f"La ruta base '{ruta_base}' no existe.")
        if not os.path.exists(ruta_destino):
            os.makedirs(ruta_destino)  # Crea la carpeta destino si no existe

        # Itera sobre todos los archivos en la carpeta base
        for nombre_archivo in os.listdir(ruta_base):
            if nombre_archivo.endswith('.json'):
                nombre_carpeta = nombre_archivo[:-5]  # Elimina ".json"
                ruta_carpeta_destino = os.path.join(ruta_destino, nombre_carpeta)

                # Crea la carpeta si no existe
                if not os.path.exists(ruta_carpeta_destino):
                    os.makedirs(ruta_carpeta_destino)

                ruta_archivo_origen = os.path.join(ruta_base, nombre_archivo)
                ruta_archivo_destino = os.path.join(ruta_carpeta_destino, nombre_archivo)

                # Copia el archivo
                shutil.copy2(ruta_archivo_origen, ruta_archivo_destino)  # Usa copy2 para copiar metadatos

                print(f"Archivo '{nombre_archivo}' copiado a la carpeta '{nombre_carpeta}'.")

    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(f"Error general: {e}")

# Ejemplo de uso:
ruta_base = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS BASE'
ruta_destino = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS PRUEBA'

copiar_archivos_a_carpetas(ruta_base, ruta_destino)
```

**Modificación de archivos JSON (Código 3):**

    *   Ejecuta el Código 3 (`copiar_y_modificar_json`) para modificar los campos específicos en los archivos JSON. Este código combina la copia de archivos y la modificación de los campos "numFactura", "tipoNota" y "numNota" dentro de la estructura "rips".

```python
import os
import json
import shutil

def copiar_y_modificar_json(ruta_base, ruta_destino):
    # ... (documentación de la función)

    try:
        if not os.path.exists(ruta_base):
            raise FileNotFoundError(f"La ruta base '{ruta_base}' no existe.")
        if not os.path.exists(ruta_destino):
            os.makedirs(ruta_destino)

        for nombre_archivo in os.listdir(ruta_base):
            if nombre_archivo.endswith('.json'):
                nombre_carpeta = nombre_archivo[:-5]
                ruta_carpeta_destino = os.path.join(ruta_destino, nombre_carpeta)

                if not os.path.exists(ruta_carpeta_destino):
                    os.makedirs(ruta_carpeta_destino)

                ruta_archivo_origen = os.path.join(ruta_base, nombre_archivo)
                ruta_archivo_destino = os.path.join(ruta_carpeta_destino, nombre_archivo)

                shutil.copy2(ruta_archivo_origen, ruta_archivo_destino)
                print(f"Archivo '{nombre_archivo}' copiado a la carpeta '{nombre_carpeta}'.")

                # ***MODIFICACIÓN IMPORTANTE: Modificar el archivo justo después de copiarlo***
                modificar_json(ruta_archivo_destino)

    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(f"Error general: {e}")


def modificar_json(ruta_archivo):
    # ... (documentación de la función)

    try:
        with open(ruta_archivo, 'r', encoding='utf-8') as f:
            datos_json = json.load(f)

        if "rips" in datos_json and isinstance(datos_json["rips"], dict):
            datos_json["rips"]["numFactura"] = None
            datos_json["rips"]["tipoNota"] = "RS"

            try:
                num_consecutivo = int(ruta_archivo.split('-')[-1].split('.')[0])
            except ValueError:
                num_consecutivo = 1

            datos_json["rips"]["numNota"] = str(num_consecutivo)

        else:
            print(f"Advertencia: El archivo '{ruta_archivo}' no tiene la estructura 'rips' esperada.")

        with open(ruta_archivo, 'w', encoding='utf-8') as f:
            json.dump(datos_json, f, indent=4, ensure_ascii=False)

    except json.JSONDecodeError:
        print(f"Error: El archivo '{ruta_archivo}' no es un archivo JSON válido.")
    except Exception as e:
        print(f"Error al procesar '{ruta_archivo}': {e}")


# Ejemplo de uso:
ruta_base = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS BASE'
ruta_destino = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS PRUEBA'

copiar_y_modificar_json(ruta_base, ruta_destino)
```

**Búsqueda y reemplazo en archivos JSON (Código 4):**

    *   Ejecuta el Código 4 (`buscar_y_reemplazar_en_json`) para realizar la búsqueda y reemplazo en masa de valores específicos en los archivos JSON. Este código se encarga de reemplazar los valores de los campos "vrServicio", "numDocumentoIdObligado", "codPrestador" y "XmlFevFile" con los valores deseados.


```python
import os
import json

def buscar_y_reemplazar_en_json(ruta_carpeta):
    """
    Busca y reemplaza valores en campos específicos de archivos JSON dentro de un directorio y sus subdirectorios.

    Args:
        ruta_carpeta (str): La ruta a la carpeta principal que contiene los archivos JSON.
    """

    try:
        if not os.path.exists(ruta_carpeta):
            raise FileNotFoundError(f"La ruta '{ruta_carpeta}' no existe.")

        for carpeta_actual, subdirectorios, archivos in os.walk(ruta_carpeta):
            nombre_carpeta_actual = os.path.basename(carpeta_actual)

            for nombre_archivo in archivos:
                if nombre_archivo.endswith('.json'):
                    if nombre_carpeta_actual == nombre_archivo[:-5]:
                        ruta_archivo = os.path.join(carpeta_actual, nombre_archivo)
                        print(f'Procesando archivo: {ruta_archivo}')

                        try:
                            with open(ruta_archivo, 'r', encoding='utf-8') as f:
                                datos_json = json.load(f)

                            # Búsqueda y reemplazo + campos adicionales
                            def reemplazar_valores(objeto):
                                if isinstance(objeto, dict):
                                    for clave, valor in objeto.items():
                                        if clave == "vrServicio":
                                            objeto[clave] = 0
                                        elif clave == "numDocumentoIdObligado":
                                            objeto[clave] = str(1024482086)
                                        elif clave == "codPrestador":
                                            objeto[clave] = str(110013905701)
                                        elif clave == "XmlFevFile":
                                            objeto[clave] = None
                                        elif clave == "valorPagoModerador":
                                            objeto[clave] = 0
                                        elif clave == "conceptoRecaudo":
                                            objeto[clave] = "05"  # Con comillas
                                        elif isinstance(valor, (dict, list)):
                                            reemplazar_valores(valor)
                                elif isinstance(objeto, list):
                                    for elemento in objeto:
                                        reemplazar_valores(elemento)

                            reemplazar_valores(datos_json)
                            print(f"Contenido JSON modificado: {datos_json}")  # Imprime JSON modificado

                            with open(ruta_archivo, 'w', encoding='utf-8') as f:
                                json.dump(datos_json, f, indent=4, ensure_ascii=False)

                        except json.JSONDecodeError:
                            print(f"Error: El archivo '{nombre_archivo}' no es un archivo JSON válido.")
                        except Exception as e:
                            print(f"Error al procesar '{nombre_archivo}': {e}")

    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(f"Error general: {e}")

# Ejemplo de uso:
ruta_carpeta = '/content/drive/MyDrive/PYS/{{Fev_ColSan_REUNION}}/RIPS ARREGLADOS '

buscar_y_reemplazar_en_json(ruta_carpeta)
```
